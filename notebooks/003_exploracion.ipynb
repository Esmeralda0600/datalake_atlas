```python
import glob
import xarray as xr
import pandas as pd
import re
```

```python

archivos = sorted(glob.glob("../datalake/raw/ECMWF_utci_2023*_v1.1_con.nc"))

archivos_unicos = list({re.search(r"(\d{8})", f).group(1): f for f in archivos if re.search(r"(\d{8})", f)}.values())
print(f"total archivos: {len(archivos_unicos)}")

salida_anual = "../datalake/raw/ECMWF_utci_2023_anual.nc"
ds_anual = xr.open_mfdataset(archivos_unicos, combine='nested', concat_dim='time', parallel=True)
ds_anual.to_netcdf(salida_anual)

df_fp = xr.open_dataset(salida_anual) 
display(df_fp['utci'].isel(lat=0, lon=0).to_dataframe().head(73))
```
