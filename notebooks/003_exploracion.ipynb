```python
import glob
import xarray as xr
import pandas as pd
import re
```

```python

files = sorted(glob.glob("../datalake/001_raw/ECMWF_utci_2023*_v1.1_con.nc"))

files_dictionary= {}
for f in files:
    match = re.search(r"(\d{8})", f)
    
    if match:
        date = match.group(1)
        files_dictionary[date] = f

unique_files = list(files_dictionary.values())
print(f"total archivos: {len(unique_files)}")

annual = "../datalake/002_trusted/ECMWF_utci_2023_anual.nc"
ds_annual = xr.open_mfdataset(unique_files, combine='nested', concat_dim='time', parallel=True)
ds_annual.to_netcdf(annual)

```

```python

df_fp = xr.open_dataset(annual) 
df_fp['utci'].isel(lat=0, lon=0).to_dataframe()
```
